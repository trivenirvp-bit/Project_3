### OUTLIER'S TREATMENT

```python
# now to treat the outliers we have :- 
# capping :- Replacing outlier values 
# in capping all outlier values will be replaced by upper extreme or lower extreme values 
def outlier_detection(data,colname):
    q1 = data[colname].quantile(0.25)   # lower quartile
    q3 = data[colname].quantile(0.75)   # upper quartile
    iqr = q3-q1   # calculating inter quartile range 

    upper_extreme = q3 + (1.5*iqr)
    lower_extreme = q1 - (1.5*iqr)

    return print('lower_extreme:-',lower_extreme,', upper_extreme:-',upper_extreme,', q1:-',q1,', q3:-',q3)
```

Q . what do you mean by model is OVERFITTED ?

Ans :- An overfitted model is a model that performs very well on the training data but poorly on unseen (test) data. 
This happens because the model has learned not just the general patterns in the data but also the noise and 
specific details (irrelevant or random variations) in the training data.

As we know  Decision trees are proned to be overfitted , due to its max_depth .

and max_depth , is itseld a hyperparamter as it is set by the users at which depth to prune the tree

## HYPERPARAMETER  :-
Hyperparameters are values that are set by the user before training a machine learning model. These include settings like learning rate, maximum tree depth,  They are not learned from the data but are specified manually or through optimization techniques.

#### Step 1: Understand Hyperparameters of Decision Tree Regressor

#### Some key hyperparameters to tune:

max_depth: The maximum depth of the tree. Limits overfitting by controlling complexity.


min_samples_split: The minimum number of samples required to split a node.


min_samples_leaf: The minimum number of samples required to be at a leaf node.


max_features: The maximum number of features to consider when looking for the best split.


max_leaf_nodes: Limits the number of leaf nodes.


#### Step 2: Define the Tuning Approach :- GRID SEARCH CV , RANDOM SEARCH 

## 1 . GRID SEARCH CV 
##### 1. GRID SEARCH :- is a technique used to find the best combination of hyperparameters for a machine learning model.
##### 2. CV(Cross-Validation) :- For each combination of hyperparameters, it evaluates the model's performance using cross-validation to ensure the results are reliable and not dependent on a single train-test split.

# 7. REGULARIZATION TECHNIQUES
##### Regularization is a technique used in machine learning to prevent overfitting by adding a penalty (or constraint) to the loss function of a model. This encourages the model to be simpler and less sensitive to noise in the data, leading to better generalization on unseen data.
##### 1. Prevents Overfitting
##### 2. Improves Generalization
##### 3.Feature Selection (L1)
##### 4. Controls Model Complexity

##### The regularization strength (λ) is a hyperparameter that controls the balance between the regularization penalty and the original error term.

## 1. L1 Regularization (Lasso):

```python
from sklearn.linear_model import LinearRegression, Lasso
```

```python
x
```

```python
y
```

```python
# after lasso, train accuracy will decrease but test accuracy will increase
l1 = Lasso() # alpha = 0.1 by default
l1.fit(x_train,y_train)
l1.score(x_train,y_train),l1.score(x_test,y_test)
# underfitted model
```

# **Hyperparameter Tuning**

```python
for i in range(100,200): # try 50 to 100, 100 to 200
  l1 = Lasso(alpha = i)
  l1.fit(x_train,y_train)
  print(f"Alpha: {i} Train: {l1.score(x_train,y_train)} Test: {l1.score(x_test,y_test)}")
```

```python
# as alpha increases train accuracy decreases and test accuracy increases
# so try big alpha value
# in above code try changing range from 50 to 100, 150 to 200
l1 = Lasso(alpha = 175) # select any value where train and test accuracy is same
l1.fit(x_train,y_train)
l1.score(x_train,y_train),l1.score(x_test,y_test)
```
